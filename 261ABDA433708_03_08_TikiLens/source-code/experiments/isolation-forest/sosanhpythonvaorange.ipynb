```python
from pathlib import Path

# Chay notebook voi working directory la thu muc chua notebook.
PROJECT_ROOT = Path.cwd().parents[2]

import pandas as pd
from sklearn.ensemble import IsolationForest


df = pd.read_csv(PROJECT_ROOT / "dataset" / "tikilens_vn.clean_products.csv")
features = ["price", "discount_rate", "quantity_sold", "revenue_estimate", "review_count"]
X = df[features]
iso_forest = IsolationForest(contamination=0.02, random_state=42)
df["anomaly_score"] = iso_forest.fit_predict(X)
df["Outlier"] = df["anomaly_score"].apply(lambda x: "Yes" if x == -1 else "No")


outliers_df = df[df["Outlier"] == "Yes"]
print(f"Tổng số sản phẩm bất thường phát hiện: {len(outliers_df)}")
print("\nDanh sách Top sản phẩm bất thường:")
print(outliers_df[["product_id", "name", "price", "quantity_sold", "revenue_estimate", "review_count"]].head())


outliers_df.to_csv(PROJECT_ROOT / "reports" / "experiments" / "isolation-forest" / "python_outliers_result.csv", index=False)
print("\n--> Đã xuất file 'python_outliers_result.csv' thành công!")
pd.set_option('display.max_columns', None)
pd.set_option('display.width', 1000)


```
